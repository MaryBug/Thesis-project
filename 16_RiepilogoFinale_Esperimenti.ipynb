```python
from pathlib import Path

import pandas as pd
import numpy as np


PROJECT_DIR = Path(
    r"C:\Users\acer\Desktop\ProgettoTesi"
)

RESULTS_DIR = (
    PROJECT_DIR
    / "risultati"
)


print("PROJECT_DIR:", PROJECT_DIR)
print("RESULTS_DIR:", RESULTS_DIR)

print("\nEsiste PROJECT_DIR:", PROJECT_DIR.exists())
print("Esiste RESULTS_DIR:", RESULTS_DIR.exists())
```

16.1 — Cartelle presenti nei risultati

```python
result_folders = sorted(
    [
        p
        for p in RESULTS_DIR.iterdir()
        if p.is_dir()
    ],
    key=lambda x: x.name.lower()
)


print(
    "Cartelle presenti in risultati:",
    len(result_folders)
)

for folder in result_folders:
    print("-", folder.name)
```

16.2 — Ricerca dei CSV dei notebook finali

```python
KEYWORDS = [
    "dn4",
    "nrs",
    "bpi",
    "riepilogo",
    "confronto",
    "mil",
    "xgboost",
    "random",
    "profil",
    "behavior",
    "pain",
    "matched",
    "cluster"
]

all_csv = list(
    RESULTS_DIR.rglob("*.csv")
)

relevant_csv = []


for path in all_csv:
    text = str(path).lower()
    if any(
        keyword in text
        for keyword in KEYWORDS
    ):
        relevant_csv.append(
            {
                "folder": path.parent.name,
                "file": path.name,
                "relative_path": str(
                    path.relative_to(
                        RESULTS_DIR
                    )
                )
            }
        )

relevant_csv_df = (
    pd.DataFrame(
        relevant_csv
    )
    .sort_values(
        [
            "folder",
            "file"
        ]
    )
    .reset_index(drop=True)
)

print(
    "CSV potenzialmente utili:",
    len(relevant_csv_df)
)

display(
    relevant_csv_df
)
```

16.3 — Ricerca mirata dei risultati finali

```python
FINAL_KEYWORDS = [
    # Notebook 9
    "permutation",
    "dn4",

    # Notebook 12
    "riepilogo_finale_dn4_nrs_bpi",

    # Notebook 13
    "xgboost",
    "random_forest",
    "rf_",
    "confronto",

    # Notebook 14
    "mil_dn4",
    "confronto_finale_dn4",

    # Notebook 15
    "duration_matched_primary_results",
    "duration_matched_sensitivity_results",
    "stabilita_ari_cluster_pain_related",
    "copertura_cluster_pain_related"
]


target_rows = []

for path in all_csv:
    text = str(path).lower()
    if any(
        keyword.lower() in text
        for keyword in FINAL_KEYWORDS
    ):
        target_rows.append(
            {
                "folder": path.parent.name,
                "file": path.name,
                "relative_path": str(
                    path.relative_to(
                        RESULTS_DIR
                    )
                )
            }
        )

target_csv_df = (
    pd.DataFrame(target_rows)
    .drop_duplicates()
    .sort_values(
        [
            "folder",
            "file"
        ]
    )
    .reset_index(drop=True)
)

print(
    "File candidati finali:",
    len(target_csv_df)
)

display(
    target_csv_df
)
```

16.4 — Ricerca dei file finali noti

```python
KNOWN_FINAL_FILES = [
    # Notebook 12
    "riepilogo_finale_DN4_NRS_BPI.csv",

    # Notebook 14
    "MIL_DN4_stabilita_5_seed.csv",
    "MIL_DN4_fold_results_5_seed.csv",
    "MIL_DN4_stabilita_statistiche.csv",
    "confronto_finale_DN4_RF_XGBoost_MIL.csv",

    # Notebook 15
    "valutazione_numero_cluster_pain_related.csv",
    "stabilita_ARI_cluster_pain_related.csv",
    "copertura_cluster_pain_related_patient_level.csv",
    "duration_matched_primary_results.csv",
    "duration_matched_sensitivity_results.csv",
]

known_files_rows = []

for filename in KNOWN_FINAL_FILES:
    matches = list(
        RESULTS_DIR.rglob(filename)
    )
    if len(matches) == 0:
        known_files_rows.append(
            {
                "requested_file": filename,
                "found": False,
                "matches": 0,
                "relative_path": None
            }
        )
    else:
        for match in matches:
            known_files_rows.append(
                {
                    "requested_file": filename,
                    "found": True,
                    "matches": len(matches),
                    "relative_path": str(
                        match.relative_to(
                            RESULTS_DIR
                        )
                    )
                }
            )

known_files_df = pd.DataFrame(
    known_files_rows
)

display(
    known_files_df
)
```
