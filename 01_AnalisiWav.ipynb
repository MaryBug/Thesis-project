```python
from pathlib import Path
import pandas as pd
import soundfile as sf

AUDIO_DIR = Path("file_wav")

wav_files = sorted(AUDIO_DIR.rglob("*.wav"))

rows = []

for wav_path in wav_files:
    try:
        info = sf.info(wav_path)

        rows.append({
            "file": wav_path.name,
            "percorso": str(wav_path),
            "durata_secondi": round(info.duration, 2),
            "sample_rate": info.samplerate,
            "canali": info.channels,
            "frames": info.frames,
            "formato": info.format,
            "subtype": info.subtype,
            "errore": None
        })

    except Exception as e:
        rows.append({
            "file": wav_path.name,
            "percorso": str(wav_path),
            "durata_secondi": None,
            "sample_rate": None,
            "canali": None,
            "frames": None,
            "formato": None,
            "subtype": None,
            "errore": str(e)
        })

df_audio = pd.DataFrame(rows)

print("Numero di file WAV trovati:", len(df_audio))
display(df_audio)
```

```python
#controllo se ci sono duplicati 
import re

def estrai_id(nome_file):
    match = re.search(r"ID(\d+)", nome_file, re.IGNORECASE)
    return int(match.group(1)) if match else None

df_audio["id"] = df_audio["file"].apply(estrai_id)

duplicati = (
    df_audio[df_audio.duplicated("id", keep=False)]
    .sort_values(["id", "file"])
)

display(duplicati)
```
