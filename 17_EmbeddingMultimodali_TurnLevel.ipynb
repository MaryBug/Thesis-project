17.16 — Funzioni embedding Whisper turn-level

```python
import soundfile as sf
import numpy as np


WHISPER_SR = 16000
WHISPER_MAX_SECONDS = 30.0
WHISPER_MAX_SAMPLES = int(
    WHISPER_SR
    * WHISPER_MAX_SECONDS
)


def load_audio_file(path):

    audio, sr = sf.read(
        str(path),
        dtype="float32"
    )

    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    if sr != WHISPER_SR:
        raise ValueError(
            f"Sample rate inatteso: {sr}"
        )

    return audio


def split_audio_30s(audio):

    return [
        audio[start:start + WHISPER_MAX_SAMPLES]
        for start in range(
            0,
            len(audio),
            WHISPER_MAX_SAMPLES
        )
    ]


def whisper_embedding_batch(
    audio_list
):

    inputs = (
        whisper_feature_extractor(
            audio_list,
            sampling_rate=WHISPER_SR,
            return_tensors="pt"
        )
    )


    input_features = (
        inputs["input_features"]
        .to(AUDIO_DEVICE)
    )


    with torch.inference_mode():

        encoder_output = (
            whisper_model.encoder(
                input_features=input_features,
                return_dict=True
            )
            .last_hidden_state
        )


    hidden_length = (
        encoder_output.shape[1]
    )


    embeddings = []


    for i, audio in enumerate(
        audio_list
    ):

        # Whisper produce circa 50 frame encoder/sec.
        # Determiniamo quanti frame corrispondono
        # realmente all'audio, escludendo il padding.

        valid_ratio = min(
            len(audio)
            / WHISPER_MAX_SAMPLES,
            1.0
        )


        valid_frames = max(
            1,
            int(
                np.ceil(
                    hidden_length
                    * valid_ratio
                )
            )
        )


        emb = (
            encoder_output[
                i,
                :valid_frames,
                :
            ]
            .mean(dim=0)
            .cpu()
            .numpy()
        )


        embeddings.append(
            emb
        )


    return np.vstack(
        embeddings
    )
```

 17.19 — Estrazione completa embedding audio Whisper-small con checkpoint e possibilità di resume

```python
from tqdm.auto import tqdm
import numpy as np
import time


AUDIO_EMBEDDING_PATH = (
    EMBEDDING_DIR
    / "embedding_audio_whisper_small_turn_level.npy"
)

AUDIO_PROGRESS_PATH = (
    EMBEDDING_DIR
    / "embedding_audio_whisper_small_progress.csv"
)


N_TURNS = len(multimodal_df)
EMBED_DIM = whisper_model.config.d_model
BATCH_SIZE = 4
CHECKPOINT_EVERY = 100


# ------------------------------------------------------------
# Eventuale resume
# ------------------------------------------------------------

if AUDIO_EMBEDDING_PATH.exists():

    audio_embeddings = np.load(
        AUDIO_EMBEDDING_PATH
    )

    if audio_embeddings.shape != (
        N_TURNS,
        EMBED_DIM
    ):
        raise ValueError(
            "Il checkpoint esistente ha shape inattesa: "
            f"{audio_embeddings.shape}"
        )

    print(
        "Checkpoint trovato."
    )

else:

    audio_embeddings = np.full(
        (
            N_TURNS,
            EMBED_DIM
        ),
        np.nan,
        dtype=np.float32
    )

    print(
        "Nuova estrazione."
    )


already_done = (
    ~np.isnan(
        audio_embeddings
    ).any(axis=1)
)


print(
    "Già completati:",
    int(already_done.sum()),
    "/",
    N_TURNS
)
```

17.20 — Estrazione

```python
def embedding_for_long_audio(audio):

    chunks = split_audio_30s(audio)
    chunk_embeddings = []
    chunk_weights = []

    for chunk in chunks:

        emb = whisper_embedding_batch(
            [chunk]
        )[0]

        chunk_embeddings.append(
            emb
        )

        chunk_weights.append(
            len(chunk)
        )

    chunk_embeddings = np.vstack(
        chunk_embeddings
    )

    chunk_weights = np.asarray(
        chunk_weights,
        dtype=float
    )

    return np.average(
        chunk_embeddings,
        axis=0,
        weights=chunk_weights
    ).astype(np.float32)


start_total = time.time()

pending_indices = np.where(
    ~already_done
)[0]


processed_since_save = 0


with tqdm(
    total=N_TURNS,
    initial=int(already_done.sum()),
    desc="Whisper turn-level"
) as pbar:

    position = 0

    while position < len(pending_indices):

        idx = pending_indices[position]

        duration = multimodal_df.loc[
            idx,
            "turn_duration_seconds"
        ]


        # ----------------------------------------------------
        # Caso raro: turno > 30 secondi
        # ----------------------------------------------------

        if duration > 30:

            audio = load_audio_file(
                multimodal_df.loc[
                    idx,
                    "audio_path_turn"
                ]
            )

            audio_embeddings[idx] = (
                embedding_for_long_audio(
                    audio
                )
            )

            position += 1
            processed_since_save += 1
            pbar.update(1)


        # ----------------------------------------------------
        # Turni <= 30 s: batch
        # ----------------------------------------------------

        else:

            batch_indices = []

            while (
                position < len(pending_indices)
                and
                len(batch_indices) < BATCH_SIZE
            ):

                candidate_idx = (
                    pending_indices[
                        position
                    ]
                )

                candidate_duration = (
                    multimodal_df.loc[
                        candidate_idx,
                        "turn_duration_seconds"
                    ]
                )

                if candidate_duration > 30:
                    break

                batch_indices.append(
                    candidate_idx
                )

                position += 1


            if len(batch_indices) == 0:
                continue


            batch_audio = [
                load_audio_file(
                    multimodal_df.loc[
                        i,
                        "audio_path_turn"
                    ]
                )
                for i in batch_indices
            ]


            batch_embeddings = (
                whisper_embedding_batch(
                    batch_audio
                )
            )


            for i, emb in zip(
                batch_indices,
                batch_embeddings
            ):

                audio_embeddings[i] = (
                    emb.astype(
                        np.float32
                    )
                )


            processed_since_save += (
                len(batch_indices)
            )

            pbar.update(
                len(batch_indices)
            )


        # ----------------------------------------------------
        # Checkpoint
        # ----------------------------------------------------

        if (
            processed_since_save
            >= CHECKPOINT_EVERY
        ):

            np.save(
                AUDIO_EMBEDDING_PATH,
                audio_embeddings
            )

            completed_now = (
                ~np.isnan(
                    audio_embeddings
                ).any(axis=1)
            )

            pd.DataFrame(
                {
                    "turn_id":
                        multimodal_df[
                            "turn_id"
                        ],

                    "completed":
                        completed_now
                }
            ).to_csv(
                AUDIO_PROGRESS_PATH,
                index=False
            )

            processed_since_save = 0


# ------------------------------------------------------------
# Salvataggio finale
# ------------------------------------------------------------

np.save(
    AUDIO_EMBEDDING_PATH,
    audio_embeddings
)


completed_final = (
    ~np.isnan(
        audio_embeddings
    ).any(axis=1)
)


pd.DataFrame(
    {
        "turn_id":
            multimodal_df[
                "turn_id"
            ],

        "completed":
            completed_final
    }
).to_csv(
    AUDIO_PROGRESS_PATH,
    index=False
)


elapsed_total = (
    time.time()
    - start_total
)


print("\nEstrazione terminata.")

print(
    "Completati:",
    int(completed_final.sum()),
    "/",
    N_TURNS
)

print(
    "Tempo totale:",
    round(
        elapsed_total / 60,
        1
    ),
    "minuti"
)

print(
    "File:",
    AUDIO_EMBEDDING_PATH
)
```

17.22 — Early fusion audio + testo

```python
from sklearn.preprocessing import normalize
import numpy as np


# Controllo allineamento
assert audio_embeddings_final.shape[0] == text_embeddings.shape[0]
assert audio_embeddings_final.shape[0] == len(multimodal_df)


# Normalizzazione L2 separata delle due modalità
audio_embeddings_l2 = normalize(
    audio_embeddings_final,
    norm="l2"
).astype(np.float32)


text_embeddings_l2 = normalize(
    text_embeddings,
    norm="l2"
).astype(np.float32)

# Equal-weight early fusion
# Ogni blocco viene moltiplicato per 1/sqrt(2): in questo modo audio e testo contribuiscono in modo bilanciato alla norma complessiva.
MODALITY_WEIGHT = 1 / np.sqrt(2)


fusion_embeddings = np.concatenate(
    [
        MODALITY_WEIGHT * audio_embeddings_l2,
        MODALITY_WEIGHT * text_embeddings_l2
    ],
    axis=1
).astype(np.float32)


print(
    "Audio:",
    audio_embeddings_l2.shape
)

print(
    "Testo:",
    text_embeddings_l2.shape
)

print(
    "Fusion:",
    fusion_embeddings.shape
)


print(
    "\nNaN fusion:",
    int(np.isnan(fusion_embeddings).sum())
)

print(
    "Inf fusion:",
    int(np.isinf(fusion_embeddings).sum())
)
```

17.23 — QC normalizzazione e fusione

```python
audio_norms = np.linalg.norm(
    audio_embeddings_l2,
    axis=1
)

text_norms = np.linalg.norm(
    text_embeddings_l2,
    axis=1
)

fusion_norms = np.linalg.norm(
    fusion_embeddings,
    axis=1
)


print(
    "Norma audio media:",
    audio_norms.mean()
)

print(
    "Norma testo media:",
    text_norms.mean()
)

print(
    "Norma fusion media:",
    fusion_norms.mean()
)


print(
    "\nRange norma fusion:",
    fusion_norms.min(),
    "-",
    fusion_norms.max()
)
```
